# Video_2 => List comprehension and function 

```python
numbers = [1,2, 3,4,5]
squared_nums = [x**2 for x in numbers]
print("squared_nums - ", squared_nums)
even_nums = [x for x in numbers if x % 2 == 0]
print("even_nums - ", even_nums)
```

```python
names = ["Abul", "Kabul", "Babul"]
ages = [25, 30, 28]
name_age_tuple = [(name, age) for name, age in zip(names,ages)] # zip helps to work with 2 lists together
print("name_age_tuple -", name_age_tuple)
```

# Video_3 => Data analysis with Pandas

```python
import pandas as pd
# Series is a sequencial data structure like array, tuple
data = [10, 20, 30, 40, 50]
series = pd.Series(data)  # creating series from list
print(series)
```

```python
# creating series with custom index
index = ['a', 'b', 'c', 'd', 'e']
series_with_index = pd.Series(data, index=index)
print(series_with_index)
```

```python
# Accessing elements in a series
print(series[0]) # value of index "0"
print(series_with_index['b']) # value of index "b"
```
