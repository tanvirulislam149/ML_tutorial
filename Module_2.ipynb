# Video_2 => List comprehension and function 

```python
numbers = [1,2, 3,4,5]
squared_nums = [x**2 for x in numbers]
print("squared_nums - ", squared_nums)
even_nums = [x for x in numbers if x % 2 == 0]
print("even_nums - ", even_nums)
```

```python
names = ["Abul", "Kabul", "Babul"]
ages = [25, 30, 28]
name_age_tuple = [(name, age) for name, age in zip(names,ages)] # zip helps to work with 2 lists together
print("name_age_tuple -", name_age_tuple)
```

# Video_3 => Data analysis with Pandas

```python
import pandas as pd
# Series is a sequencial data structure like array, tuple
data = [10, 20, 30, 40, 50]
series = pd.Series(data)  # creating series from list
print(series)
```

```python
# creating series with custom index
index = ['a', 'b', 'c', 'd', 'e']
series_with_index = pd.Series(data, index=index)
print(series_with_index)
```

```python
# Accessing elements in a series
print(series[0]) # value of index "0"
print(series_with_index['b']) # value of index "b"
```

```python
# Performing operations on a Series
print(series * 2)
print(series_with_index + 5)
print(series[3] * 2)
print(series_with_index['b'] + 5)
```

```python
# Filtering a Series
print(series[series > 30])

# Statistical operations on Series
print(series.mean())
print(series.std())
```

```python
# Useful methods and attributes
print(series.index)
print(series.values)
print(series.describe())
print(series_with_index.index)
print(series_with_index.values)
print(series_with_index.describe())
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 45, 43],
    "City": ["Dhaka", "Ctg", "Dhaka", "Khulna"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Basic statistics
print(df.describe())

# Accessing specific column
print("\nAges: ")
print(df["Age"])

# Filtering data
print("\nPeople older than 30 -")
print(df[df['Age'] > 30])
```

```python
# Grouping data
print(df["Age"].mean()) # shows mean of one column
print("\n Average age by city: ")
print(df.groupby("City")['Age'].mean()) # mean of ages after grouping by city name
```

```python
# Sorting data
print("\n Sort by age")
print(df.sort_values("Age"))
```

```python
df.columns = ["num_value", "station", "year", "month", "day", "rainfall", "sunshine", "humidity", "temperature"]  # can change the column names like this
```

```python
temp = df["temperature"]  # shows only one column in series
temp
```

```python
df1 = df[["temperature", "humidity"]] # makes new dataframe with just two column
df1
```

```python
df2 = df.iloc[0:10, [1,2]]  # shows 0-10 rows and 1st and 2nd column
df2
```

```python
# df.iloc works on column index and df.loc works on column name
df3 = df.iloc[1::2, [0, 1, 3]] # shows odd numbered rows starting from one
df3
```

```python
temp = df["temperature"]
print("Mean", temp.mean())
print("Standard Daviation", temp.std())
print("Variance", temp.var())
print("Lower Quartile", temp.quantile(0.25))
print("Median", temp.quantile(0.50))
print("Median", temp.median())
print("Upper Quartile", temp.quantile(0.75))
print("Skewness", temp.skew())
print("Kurtosis", temp.kurt())
print("Min", temp.min())
print("Max", temp.max())
```
